```python
# Importing library
import spacy
```

```python
# Selecting Model
spacy.cli.download('en_core_web_lg')
nlp = spacy.load('en_core_web_lg')
```

```python
# nlp = spacy.load('en_core_web_sm')  # model for NLP # previous model
```

```python
# reading the small english language
file_path = '/Cover letter.txt'
with open(file_path, 'r') as file:
    text = file.read()
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
# processing the text using spacy
doc = nlp(text)
print(doc)
```

```python
#extracting the small language models
print('Named Entities:')
for ent in doc.ents:
    print(f'{ent.text} - {ent.label_}')
```
