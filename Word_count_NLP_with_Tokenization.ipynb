```python
# Load text from the file and tokenize the text nltk is library for nlp
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from collections import Counter
import matplotlib.pyplot as plt

nltk.download('stopwords')
nltk.download('punkt_tab')



```

```python
# Reading the small english language model
file_path = '/content/Cover letter.txt'
with open(file_path, 'r') as file:
    text = file.read()
tokens=word_tokenize(text)
```

```python
stop_word = set(stopwords.words('english'))
filtered_tokens = [word.lower() for word in tokens if word.isalpha() and word.lower() not in stop_word]
word_freq = Counter(filtered_tokens)
```

```python
# EXtract the top 10 most frequest word
top_words = word_freq.most_common(10)
word,count = zip(*top_words)
```

```python
# visulaising the bar chart
plt.figure(figsize=(10, 6))
plt.bar(word, count)
plt.xlabel('Word')
plt.ylabel('Frequency')
plt.title('Top 10 Most Frequent Words')
plt.xticks(rotation=45)
plt.show()
```
